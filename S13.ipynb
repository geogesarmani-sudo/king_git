```python
# Exercice 9
# Moyenne et écart-type pour l'Afrique
afrique = ONU.loc[ONU["continent"] == "Africa", "lifeExp"]
print("Afrique - Moyenne :", afrique.mean())
print("Afrique - Écart-type :", afrique.std())

# Moyenne et écart-type pour l'Europe
europe = ONU.loc[ONU["continent"] == "Europe", "lifeExp"]
print("Europe - Moyenne :", europe.mean())
print("Europe - Écart-type :", europe.std())
# En Europe, l’écart‑type est faible , la moyenne reflète bien la réalité.

# En Afrique, l’écart‑type est plus fort , la moyenne cache des inégalités entre pays.
```

```python
# Exercice 10
extrait_2 = ONU.loc[(ONU["continent"] == "Africa") & (ONU["iso3"].isin(["COG", "COD"])),["country", "iso3", "gdpPercap","lifeExp","pop"]]

print(extrait_2)
#  L’évolution dans le temps révèle que les deux Congo ne sont pas seulement proches aujourd’hui, mais qu’ils ont progressé ensemble et que leurs écarts se sont atténués.
```

```python
# Exercice 11
# Moyenne simple : chaque pays pèse pareil
moyenne_simple = ONU["lifeExp"].mean()

# Moyenne pondérée par la population
moyenne_ponderee = (ONU["lifeExp"] * ONU["pop"]).sum() / ONU["pop"].sum()

print(f"Moyenne simple : {moyenne_simple:.2f}")
print(f"Moyenne pondérée : {moyenne_ponderee:.2f}")
#  La moyenne pondérée est plus représentative de ce que vit “l’habitant moyen” de la planète.
```

```python
# Exercice 12
correlation = ONU["gdpPercap"].corr(ONU["lifeExp"])
print(f"Corrélation : {correlation:.2f}")

# Pays atypique parmi les riches (PIB > 20 000 $)
riches = ONU[ONU["gdpPercap"] > 20000]
atypique = riches.sort_values(by="lifeExp").head(1)
print(atypique[["country", "gdpPercap", "lifeExp"]])
# Non, une corrélation forte ne prouve pas la causalité. D'autres facteurs (accès aux soins, éducation, stabilité politique, nutrition) peuvent expliquer les deux variables simultanément
```
